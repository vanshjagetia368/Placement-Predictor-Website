# NOW WE WILL DO THE EXTRACTION OF INPUT AND OUTPUT COLUMNS

```python
X=df.iloc[:,0:2] #input columns 0,1
Y=df.iloc[:,-1] #output column last one placement
```

```python
print(X)
```

```python
print(Y)
```

# AS SCALING THE VALUES IS NOT NECESSARY HERE WE WILL NOW MOVE ON TO THE TRAINING AND TESTING SPLIT OF THE MODEL

```python
from sklearn.model_selection import train_test_split
```

```python
#The syntax is train_test_split(independent_variables,dependent_variables,test_size)
#Now the test size indicates about how much data to go in the training and testing part of the model
#let say test size =0.1 which means 10 percent data into testing and 90 percent data into training
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.1)
```

```python
print(X_train)
```

```python
print(X_test)
```

```python
print(Y_train)
```

```python
print(Y_test)
```

# NOW WE WILL SCALE THE VALUES

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
```

```python
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
print(X_train)
```

```python
print(X_test)
```

#  NOW WE HAVE ALL THE THINGS TO TRAIN THE MODEL SO LTS BEGIN WITH THE TRAINING

```python
#WE WILL USE LOGISTIC REGRESSION CLASSIFIER FOR THE MODEL TRAINING
from sklearn.linear_model import LogisticRegression
```

```python
clf=LogisticRegression()
```

```python
clf.fit(X_train,Y_train)
```

```python
y_pred=clf.predict(X_test)
```

```python
print(Y_test)
```

# NOW OUR MODEL IS SUCCESSFULLY TRAINED NOW LETS FIND THE ACCURACY OF THE MODEL

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(Y_test,y_pred)
```

# NOW LET US TRY TO PLOT THE BOUNDARY THAT OUR MODEL CHOOSE TO PREDICT THE OUTPU

```python
from mlxtend.plotting import plot_decision_regions
```

```python
plot_decision_regions(X_train, Y_train.values, clf=clf, legend=2)
plt.xlabel('CGPA')
plt.ylabel('IQ')
plt.title('Logistic Regression Linear Decision Boundary')
plt.show()
```

# NOW WE WILL EXPORT THE MODEL

```python
import pickle
#pickle creates an object to a file which we can use in any other environments
```

```python
pickle.dump(clf,open('model.pkl','wb'))
pickle.dump(scaler, open("scaler.pkl", "wb"))
```
